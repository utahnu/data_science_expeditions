## HR Diagram

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("star_stats.csv")
data
```

```python
fig, ax = plt.subplots()
ax.scatter(data = data, x = "temperature", y = "luminosity")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("temperature")
ax.set_ylabel("luminosity")
ax.set_title("Temperature vs. Luminosity log-log")
plt.show()
```

```python
data["star_type"].unique()
```

```python
brown_dwarfs = data.loc[data['star_type'] == 'Brown Dwarf']
red_dwarfs = data.loc[data['star_type'] == 'Red Dwarf']
white_dwarfs = data.loc[data['star_type'] == 'White Dwarf']
main_sequences = data.loc[data['star_type'] == 'Main Sequence']
super_giants = data.loc[data['star_type'] == 'Super Giant']
hyper_giants = data.loc[data['star_type'] == 'Hyper Giant']
```

```python
fig, ax = plt.subplots(figsize = (15,10))
ax.scatter(data = brown_dwarfs, x = 'temperature', y = 'luminosity', label = 'brown_dwarf')
ax.scatter(data = red_dwarfs, x = 'temperature', y = 'luminosity', label = 'red_dwarf')
ax.scatter(data = white_dwarfs, x = 'temperature', y = 'luminosity', label = 'white_dwarf')
ax.scatter(data = main_sequences, x = 'temperature', y = 'luminosity', label = 'main_sequence')
ax.scatter(data = super_giants, x = 'temperature', y = 'luminosity', label = 'super_giant')
ax.scatter(data = hyper_giants, x = 'temperature', y = 'luminosity', label = 'hyper_giant')
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("temperature")
ax.set_ylabel("luminosity")
ax.set_title("Temperature vs. Luminosity log-log")
ax.legend()
ax.axvline(x = 17000)
plt.show()
```

```python
print(hyper_giants['luminosity'].max(), hyper_giants['luminosity'].min())
print(super_giants['luminosity'].max(), super_giants['luminosity'].min())
```
